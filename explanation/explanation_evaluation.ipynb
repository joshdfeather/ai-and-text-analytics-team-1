```python
import json
from sentence_transformers import CrossEncoder

with open('-----.json', 'r') as f: ## Add File Name Here
    data = json.load(f)

model_name = 'cross-encoder/nli-deberta-v3-base'
nli = CrossEncoder(model_name)
txt = [[row['model_explanation'], row['long_answer']] for row in data]
probs = nli.predict(txt)
predictons = probs.argmax( axis = 1)
corect = (predictons == 1).sum()
wrong = (predictons != 1).sum()

print(f"Correct: {corect}")
print(f"Wrong: {wrong}")
print(f"Accuracy:  {(corect/len(data)):.2%}")
```
